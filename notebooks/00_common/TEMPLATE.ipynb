```python
# Importar librerías estándar
import sys
from pathlib import Path
import warnings

# Datos y análisis
import numpy as np
import pandas as pd

# Visualización
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go

# Reproducibilidad
np.random.seed(42)
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)
warnings.filterwarnings('ignore')

# Detectar raíz del repo
repo_root = Path.cwd()
while repo_root != repo_root.parent:
    if (repo_root / 'pyproject.toml').exists():
        break
    repo_root = repo_root.parent

if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

print(f"✅ Entorno listo")
print(f"   Python: {sys.version.split()[0]}")
print(f"   pandas: {pd.__version__}")
print(f"   Raíz: {repo_root}")
```

## 📥 Carga de Datos

```python
# Ruta a datos
data_dir = repo_root / 'data' / 'raw'

# Cargar datasets necesarios
# NOTA: Primero ejecuta: python data/synthetic_generators/generate_cli.py
orders = pd.read_csv(data_dir / 'orders.csv')
inventory = pd.read_csv(data_dir / 'inventory.csv')

print(f"📊 Datasets cargados:")
print(f"   Orders: {len(orders):,} registros")
print(f"   Inventory: {len(inventory):,} registros")
```

```python
def print_summary(df, name="DataFrame"):
    """Print data summary for quick inspection."""
    print(f"\n{name}:")
    print(f"  Shape: {df.shape}")
    print(f"  Memory: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
    print(f"  Missing: {df.isnull().sum().sum()} values")
    return df
```
